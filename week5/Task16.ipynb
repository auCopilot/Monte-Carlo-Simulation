```python
import numpy as np
def queue_antithetic(n_costumers):

    # Antithetic variables
    n = n_costumers
    U = np.random.uniform(0, 1, n)
    V = np.random.uniform(0, 1, n)

    A = -1.1*np.log(U)
    S = -np.log(V)

    A1 = -1.1*np.log(V)
    S1 = -np.log(U)

    A2 = -1.1*np.log(1-U)
    S2 = -np.log(1-V)


    def antithetic_var(B,B_prime):
        # Computes the variance of mean of samples
        return np.var(0.5*(B + B_prime))


    B = np.empty(n)

    B_1 = np.empty(n)

    B_2 = np.empty(n)


    for i in range(n):

        B[i] = max(B[i-1] + S[i-1] - A[i],0) if i > 0 else 0
        B_1[i] = max(B_1[i-1] + S1[i-1] - A1[i],0) if i > 0 else 0
        B_2[i] = max(B_2[i-1] + S2[i-1] - A2[i],0) if i > 0 else 0

    anti_var1 = antithetic_var(B,B_1)
    anti_var2 = antithetic_var(B,B_2)
    normal_var = np.var(B)
    return anti_var1, anti_var2, normal_var

n_costumers = 20
N_sim = 1000
sims = np.empty((3,N_sim))
for N in range(N_sim):

    anti_var1, anti_var2, normal_var = queue_antithetic(n_costumers)

    sims[:, N] = np.array([anti_var1, anti_var2, normal_var])
print("The variance of Antithetic 1: ", anti_var1)
print("The variance of Antithetic 2: ", anti_var2)
print("The variance of Crude sampling: ", normal_var)

means = np.mean(sims, axis=1)
print("The mean variance of Antithetic 1: ", means[0])
print("The mean variance of Antithetic 2: ", means[1])
print("The mean variance of Crude sampling: ", means[2])


```

We see that it does indeed reduce variance
