```python
import numpy as np
def queue_antithetic(wait_time):

    # Antithetic variables
    n = wait_time
    U = np.random.uniform(0, 1, n)
    V = np.random.uniform(0, 1, n)

    A = -1.1*np.log(U)
    S = -np.log(V)

    A1 = -1.1*np.log(V)
    S1 = -np.log(U)

    A2 = -1.1*np.log(1-U)
    S2 = -np.log(1-V)

    B = np.empty(n)
    B_1 = np.empty(n)
    B_2 = np.empty(n)


    for i in range(n):

        B[i] = max(B[i-1] + S[i-1] - A[i],0) if i > 0 else 0
        B_1[i] = max(B_1[i-1] + S1[i-1] - A1[i],0) if i > 0 else 0
        B_2[i] = max(B_2[i-1] + S2[i-1] - A2[i],0) if i > 0 else 0



    return B[-1], B_1[-1], B_2[-1]

def queue_crude(wait_time):

    # Antithetic variables
    n = wait_time
    U = np.random.uniform(0, 1, n)
    V = np.random.uniform(0, 1, n)

    A = -1.1*np.log(U)
    S = -np.log(V)
    B = np.empty(n)
    for i in range(n):

        B[i] = max(B[i-1] + S[i-1] - A[i],0) if i > 0 else 0

    return B[-1]


def antithetic_var(B,B_prime):
        # Computes the variance of mean of samples
        m = len(B)
        return np.var(0.5*(B + B_prime) ) / m

wait_time = 100
N_sim = 10000
N_sim_crude = 2*N_sim

simulations = np.array([list( queue_antithetic(wait_time) ) for _ in range(N_sim)])
expectations = [
    np.mean(0.5*(simulations[:, 0] + simulations[:, 1]), axis=0),
    np.mean(0.5*(simulations[:, 0] + simulations[:, 2]), axis = 0)
                ]

crude = [queue_crude(wait_time) for _ in range(N_sim_crude)]
expectations = np.append(expectations, np.mean(crude))


variances = [
            antithetic_var(simulations[:,0], simulations[:,1]),
            antithetic_var(simulations[:,0], simulations[:,2]),
            np.var(crude) / N_sim_crude
             ]

```
