To find the theta that minimizes the function, we find a root of the derivative of H

```python
import numpy as np

from week10 import *

N = 100
beta = np.array([1/2, 1, 3/2])
def sampler():
    X = np.random.gamma( shape = 2,
                     scale= beta[:, None],
                     size= (3, N)
                     )
    return X

first_max = lambda x, theta: (x[0] * (theta * x[0] + x[1] > (1 - theta) * x[2]))

second_max = lambda x, theta: (x[2] * (theta * x[0] + x[1] < (1 - theta) * x[2]))

# Each row is with respect to a new theta in [0, 0.1, ..., 0..9]
dH = lambda x, theta: first_max(x, theta) - second_max(x, theta)

a = [0.5/n**0.8 for n in range(1, 102)]

robbins_monro(func= dH, sampler= sampler, a = a, theta0 = 0.5, max_iter= 100)
```
