We simulate $N=10^5$ samples of the standard normal distribution and find the proportion of samples that are above the approximate 95th percentile of the standard normal distribution.

```python
import numpy as np
import matplotlib.pyplot as plt


N = 10**5
z = np.random.normal(0, 1, N)
q = 1.645

print(f"Proportion: {np.sum(z > q) / N:.4f}")
```

We can then find the mean of the samples that are above the approximate 95th percentile of the standard normal distribution. I denote them as $Y ~ Z | Z > 1.645$.

```python
y = z[z > q]
print(f"Mean of Y: {np.mean(y):.4f}")
```

We are sampling from $Y$, therefore $\hat Y = \dfrac{1}{N} \sum_{i=1}^{N} Y_i $ by LLN converges in probability to $E[Y]$. Thus:
$$
\dfrac{1}{N} \sum_{i=1}^{N} Y_i \xrightarrow{p} E[Y] = E[Z | Z > 1.645]
$$

It follow that we can then also estimate $E[y]$ by monte carlo integrtion.

```python
from scipy import stats
import numpy as np

N = 10**5
alpha = 0.05

u = np.random.uniform(0, alpha, N)

z_hat = np.sum(stats.norm.ppf(1-u)) / N
z_hat
```
