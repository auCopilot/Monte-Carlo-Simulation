We want top use the Asmussen-Kroese method to find the probability of
$$
P(X_1 + X_2 > 50)
$$
Now we recall that the law of total expectation states that
$$
E[X] = E[E[X|Y]]
$$

We can use that
$$
P(S_n > x, M_n = X_n | X_1, ..., X_{n-1}) = E[I(S_n > x, M_n = X_{n-1}) | X_1, ..., X_{n-1}] = \bar F(max(M_{n-1}, x - S_{n-1}))
$$
Then the law of interated expectation gives us
$$
E[E[I(S_n > x, M_n = X_n) | X_1, ..., X_{n-1}]] = E[\bar F(max(M_{n-1}, x - S_{n-1}))] =$$
$$
 E[I(S_n > x, M_n = X_n)] = P(S_n > x, M_n = X_n)
$$
Now we use the result from excercise 28, so we can write
$$
P(S_n > x) = n P(S_n > x, M_n = X_n) = n E[\bar F(max(M_{n-1}, x - S_{n-1}))]
$$
Thus we can estimate $P(S_n > x)$ by simulating $N$ samples of $(X_1, ..., X_{n-1})$ and computing the average of $\bar F(max(M_{n-1}, x - S_{n-1}))$ and multiplying by $n$.

```python
from scipy.stats import pareto
import numpy as np
def F_bar(x):
    """tail function of Pareto shape = 2, scale = 1"""
    return pareto.sf(x, 2, scale=1)

N = 100000
n = 2
x = 50
# 10000 Column samples of (X1,X2)
X = pareto.rvs(2, scale=1, size=(2, N))

# M_n-1 = max(X1) = X1
M = np.max(X[0:n-1, :].reshape(1,-1), axis = 0)
# S_n-1 = sum(X1) = X1
S= np.sum(X[:n-1, :], axis=0, keepdims=True)

max_val = np.maximum(M, x - S)

F_bars = F_bar(max_val)

l = n * np.mean(F_bars, axis = 1)

print(f"Estimated probability P(X1 + X2 > {x}) = {l[0]}")

# Crude Monte Carlo simulation to estimate P(X1 + X2 > 50)
N = 10000000
n = 2
x = 50
# 10000 Column samples of (X1,X2)
X = pareto.rvs(2, scale=1, size=(2, N))

bool = np.sum(X, axis=0) > x
l_crude = np.mean(bool)
print(f"Estimated probability P(X1 + X2 > {x}) = {l_crude}")
```
