# (2)
We define the transistion matrix and find the left eigenvector

```python
import numpy as np


N = 11

connections = {0 : None,
               1 : [2],
               2 : [1],
               3 : [0,1],
               4 : [3, 5],
               5 : [4, 1],
               6 : [1,4],
               7 : [1,4],
               8 : [1,4],
               9 : [5],
               10 : [5]}
def define_probabilities(state_connections):

    P = np.zeros((len(state_connections), len(state_connections)))

    for k, v in state_connections.items():
        if not v:
            P[k, : ] = np.repeat(1 / N, N)
        else:
            for i in range(N):
                if i in v:
                    P[k, i] = 0.85 / len(v) + 0.15 / N
                else:
                    P[k, i] = 0.15 / N
    return P

Pmat = define_probabilities(connections)

# Left eigen vectors
eigenvalues, eigenvectors = np.linalg.eig(Pmat.T)

# Select the one corresponding to lambda = 1
idx = np.argmin(np.abs(eigenvalues - 1))

pi = eigenvectors[:, idx].real

# Normalize to sum to one
pi /= np.sum(pi)

# Sample
N_samples = 10**5

current_state = 5
X = np.zeros(N_samples, dtype=int)

def markov_sample(current_state, p_mat):

    return int(np.random.choice(
        np.arange(len(p_mat)),
        p=p_mat[current_state]
    ))

for i in range(N_samples):
    X[i] = markov_sample(current_state, Pmat)
    current_state = X[i]

# State F = 5
counter = 0
idx = np.where(X == 5)[0][0]
distances = []
for x in X[idx + 1:]:
    counter += 1
    if x == 5:
        distances.append(counter)
        counter = 0


mean_distance = np.mean(distances)

print("Mean distance:", mean_distance)
print("1  / Entry for F in first left eigenvector:", 1 / pi[5])
```
