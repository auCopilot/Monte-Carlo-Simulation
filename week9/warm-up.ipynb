```python
import numpy as np
import matplotlib.pyplot as plt
from week9 import pertubation_analysis_derivative
N = 100000
# Define the derivative
dH = lambda x: np.where(x > 1, 1, 0)

theta = np.linspace(0,10, 100)
X = np.random.normal(theta[:, None], 1, size=(len(theta), N))

l, s = pertubation_analysis_derivative(X, dH)

plt.plot(theta, l)
```

We now use the score function algorithm instead. We need to compute the score function for the N(theta, 1) distribution. The score function is given by
$$
\dfrac{\partial}{\partial \theta} \log f(x; \theta) = \dfrac{\partial}{\partial \theta}(\log(e^{-(x-\theta)^2/2}) - \log(\sqrt{2\pi}))= \dfrac{\partial}{\partial \theta}(-(x-\theta)^2/2) - \log(\sqrt{2\pi})) = x - \theta
$$

```python
score_func = lambda x, theta: x - theta

sf = lambda x: score_func(x, theta[:, None])
H = lambda x: np.maximum(x-1, 0)
from week9 import score_function_derivative

ls, ss = score_function_derivative(X, h_func= H, score_func= sf)
plt.plot(theta, ls)
```
