# Task 38: Perturbation Analysis and Score Function Derivative Estimation

We consider
$$
z(\theta) = E[X^2]
$$
where $X \sim N(\theta, 1)$. We will estimate the derivative of $z(\theta)$ with respect to $\theta$ using both perturbation analysis and the score function method and compare the variances of the two methods. First we do the straight forward score function method, as the score function is easy to compute for the normal distribution. We have
$$
\dfrac{\partial}{\partial \theta} \log f(x; \theta) = \dfrac{\partial}{\partial \theta}(\log(e^{-(x-\theta)^2/2}) - \log(\sqrt{2\pi}))= \dfrac{\partial}{\partial \theta}(-(x-\theta)^2/2) - \log(\sqrt{2\pi})) = x - \theta
$$


```python
import numpy as np
theta = 2
N = 1000
X = np.random.normal(theta, 1, size=N)
sf = lambda x: x - theta
H = lambda x: x**2
from week9 import score_function_derivative
ls, ss = score_function_derivative(X, h_func= H, score_func= sf)
print("Variance of derivative estimate:", ss**2)
```

Now we can express the derivative of $H(Z;\theta)$ with respect to $\theta$ as follows:
$$
H(Z; \theta) = (\theta + Z)^2 \implies \dfrac{\partial}{\partial \theta} H(Z; \theta) = 2(\theta + Z)
$$
then it is we can sample from $Z$ and estimate the derivative of $z(\theta)$ with respect to $\theta$ using perturbation analysis.

```python
dh = lambda z: 2*(theta + z)
Z = np.random.normal(0, 1, size=N)
from week9 import pertubation_analysis_derivative
l, s = pertubation_analysis_derivative(Z, dh)
print("Variance of derivative estimate:", s**2)

print("Relative variance of perturbation analysis to score function method:", s**2 / ss**2)
```
