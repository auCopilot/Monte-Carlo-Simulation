```python
import numpy as np
import matplotlib.pyplot as plt
# Inverse Gaussian
f = lambda x: (2*np.pi*x**3)**-0.5*np.exp(-0.5*(x-1)**2/x)
x = np.linspace(0.01,10,1000)
g = lambda x, lam : lam*np.exp(-lam*x)

plt.plot(x,f(x), c = "red")
plt.plot(x,2.7*g(x,0.5), c = "blue")
plt.show()
```

```python
def accept_reject(f, g, g_inv, C, n = 10**5):
    v = np.random.uniform(0,1,n)
    u = np.random.uniform(0,1,n)

    y = g_inv(v)
    f_star = f(y) / (C*g(y))
    return y[u < f_star]

C = 2.7
g_inv = lambda x, lam : -1/lam * np.log(x)

y = accept_reject(lambda x: f(x), lambda v : g(v,0.5), lambda v: g_inv(v,0.5), C)
```

```python
plt.hist(y, bins=50, density=True, alpha=0.5)
plt.plot(x, f(x), linewidth=2)
plt.show()
```
